```python
# import pandas for data analysis
import pandas as pd
# import seaborn for visualisations
import seaborn as sns 


# read the data from the csv file
data = pd.read_csv('loan_training_data.csv')

pattern = r'(([A-Z][A-Z])|([A-Z]))'

data['PostcodeStart'] = data.Postcode.str.extract(pattern)[0]

data
```

```python
data.groupby('Decision')['Age'].describe().round(2) 
```

```python
pd.crosstab(data['PostcodeStart'], data['Decision'], margins=True) 
```

```python
data = []

with open('loan_training_data.csv', 'r') as file:

    for i in file.read().splitlines():
        data.append(i.split(','))
    print(data)
```

```python
sns.displot(data=data, x='Age', hue='Decision', stat='proportion', multiple='fill', aspect=2);
```
